# Caracal Bench Single - GPU T4 x2

Roda **1 benchmark** isolado em GPU T4 x2. Use pra fracionar quota em runs curtos.

**Edita 2 vars no topo:**
- `BENCH` - qual rodar: `probe` / `mmlu_security` / `secqa` / `cybermetric` / `cti_bench` / `humaneval`
- `CHECKPOINT_DATASET` - slug do adapter LoRA

**Settings (sidebar):**
1. Accelerator -> GPU T4 x2 (machine_shape NvidiaTeslaT4)
2. Internet ON
3. Persistence Variables and Files

**Tempo por bench (adapter + base):**
| Bench | ETA T4 x2 |
|---|---|
| probe | ~30min |
| mmlu_security | ~20min |
| secqa | ~30min |
| cybermetric (tier 500) | ~30min |
| cti_bench (RCM 500) | ~30min |
| humaneval | ~100min |

```python
BENCH = "cybermetric"  # probe|mmlu_security|secqa|cybermetric|cti_bench|humaneval
CHECKPOINT_DATASET = "pedroafonso2/caracal-base-3b-s01"
OUTPUT_DATASET = f"pedroafonso2/caracal-bench-{BENCH.replace('_', '-')}-s01"

ALL_BENCHES = ["probe", "mmlu_security", "secqa", "cybermetric", "cti_bench", "humaneval"]
assert BENCH in ALL_BENCHES, f"BENCH must be one of {ALL_BENCHES}"
SKIP_FLAGS = [f"--skip-{b.replace('_', '-')}" for b in ALL_BENCHES if b != BENCH]
# alias: mmlu_security CLI flag is --skip-mmlu (not --skip-mmlu-security)
SKIP_FLAGS = ["--skip-mmlu" if f == "--skip-mmlu-security" else f for f in SKIP_FLAGS]
print(f"bench={BENCH} checkpoint={CHECKPOINT_DATASET}")
print(f"skip flags: {SKIP_FLAGS}")
```

```python
import subprocess

ckpt_dir = "/kaggle/working/ckpt"
subprocess.run(
    [
        "kaggle",
        "datasets",
        "download",
        "-d",
        CHECKPOINT_DATASET,
        "-p",
        ckpt_dir,
        "--unzip",
        "--force",
    ],
    check=True,
)
subprocess.run(["ls", "-la", ckpt_dir], check=True)
```

```python
import subprocess
import sys

cmd = [
    sys.executable,
    "-u",
    "eval/run_bench_all.py",
    "--adapter",
    ckpt_dir,
    "--out-dir",
    "/kaggle/working/bench-adapter",
] + SKIP_FLAGS
print("Running adapter bench:", " ".join(cmd), flush=True)
subprocess.run(cmd, check=True)
```

```python
import subprocess
import sys

cmd = [
    sys.executable,
    "-u",
    "eval/run_bench_all.py",
    "--out-dir",
    "/kaggle/working/bench-base",
] + SKIP_FLAGS
print("Running base bench:", " ".join(cmd), flush=True)
subprocess.run(cmd, check=True)
```

```python
import json
from pathlib import Path

adapter_sum = json.loads(Path("/kaggle/working/bench-adapter/summary.json").read_text())
base_sum = json.loads(Path("/kaggle/working/bench-base/summary.json").read_text())

ad_step = adapter_sum.get(BENCH, {})
ba_step = base_sum.get(BENCH, {})

diff = {
    "checkpoint_dataset": CHECKPOINT_DATASET,
    "bench": BENCH,
    "adapter": ad_step,
    "base": ba_step,
}
pub_dir = Path("/kaggle/working/bench-published")
pub_dir.mkdir(parents=True, exist_ok=True)
(pub_dir / f"bench_{BENCH}_vs_base.json").write_text(json.dumps(diff, indent=2))
print(
    json.dumps({"adapter": ad_step.get("metrics"), "base": ba_step.get("metrics")}, indent=2)[:2000]
)
```

```python
import json
import subprocess

metadata = {
    "title": f"Caracal Bench {BENCH} s01",
    "id": OUTPUT_DATASET,
    "licenses": [{"name": "Apache-2.0"}],
}
(pub_dir / "dataset-metadata.json").write_text(json.dumps(metadata, indent=2))

r = subprocess.run(
    ["kaggle", "datasets", "create", "-p", str(pub_dir), "--public"],
    capture_output=True,
    text=True,
    check=False,
)
print(r.stdout, r.stderr)
if r.returncode != 0:
    subprocess.run(
        ["kaggle", "datasets", "version", "-p", str(pub_dir), "-m", f"bench {BENCH}"], check=True
    )
print(f"Published -> {OUTPUT_DATASET}")
```
